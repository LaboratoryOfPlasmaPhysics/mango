# Getting started

MANGO is a dataset that classifies, contextualizes and spatially normalizes spacecraft observations of Earth's magnetosphere. `space_mango` is the Python client: it lists what is available and downloads the rows you ask for as polars or pandas data frames.

```{note}
These pages run on a small real-data sample served locally (dataset version `2026.0-docs-sample`). With the public server, the same code returns the full dataset.
```

````python
import space_mango as mango

mango.regions()
````

````python
mango.magnetosheath
````

````python
mango.describe("magnetosheath")
````

`R_norm` is the normalized position of a row between the region boundaries (magnetopause and bow shock for the magnetosheath), and the `*_swi` columns are local measurements expressed in the SWI frame. See the [user guide](../user_guide.md) for the full column table.

````python
mango.spacecraft("magnetosheath")
````

````python
mango.search("density")
````

````python
mango.count("magnetosheath", bz_imf_max=-2)
````

````python
r = mango.magnetosheath.get_data(columns=["Time", "Np", "Bz_imf", "R_norm"], bz_imf_max=-2)
r
````
