# Working with results

`get_data` returns a `MangoResult`: the data plus its metadata, dataset version and the query that produced it.

```{note}
These pages run on a small real-data sample served locally (dataset version `2026.0-docs-sample`). With the public server, the same code returns the full dataset.
```

````python
import space_mango as mango

query = dict(columns=["Time", "Bx", "By", "Bz", "Np", "R_norm"], spacecraft="THA", start="2008-08-01", stop="2008-09-01")
mango.count("magnetosphere", **query)
````

Check `download_mb_estimate` with `count()` before downloading.

````python
r = mango.magnetosphere.get_data(**query)
````
