# Working with results

`get_data` returns a `MangoResult`: the data plus its metadata, dataset version and the query that produced it.

```python
import space_mango as mango

r = mango.magnetosphere.get_data(columns=["Time", "Bx", "By", "Bz", "Np", "R_norm"], d_msp_min=0.8)
```

```python
r.version, len(r), r.columns
```

```python
r.metadata["Bz"]
```

```python
df = r.to_pandas()
df.head()
df.attrs["mango"]["columns"]["Np"]
```
